# apply selections

```python
%%time


JET_PT_CUT = 30
MUON_PT_CUT = 30.0


cscRechitClusterPhi0 = {}
cscRechitClusterPhi1 = {}
deltaPhi_cluster  = {}
deltaR_cluster  = {}
cscRechitClusterDNN0  = {}
cscRechitClusterDNN1  = {}
cscRechitClusterTime0  = {}
cscRechitClusterTime1  = {}
cscRechitClusterSize0  = {}
cscRechitClusterSize1  = {}
cscRechitClusterEta = {}
cscRechitClusterNStation = {}
weight = {}
HMTEff = {}
sel_ev = {}
sel = {}
pileupWeightUp= {}
pileupWeightDown= {}
scaleWeights= {}

categories = ['lowMET', 'highMET']

region_list = [ 'signal']
jetmet = {}
for jetmetsys in [0,1,2]: # 0 is nominal, 1 is up, 2 is down
    if not jetmetsys == 0:continue
    for region in region_list:
        for tree_k in tree.keys():
            T = tree[tree_k]

            if (not "signal" in region) and (not "data" in tree_k):continue
            for cat in categories:
                k = f'{region}_{tree_k}_{cat}'
                if not cat in jetmet.keys():jetmet[cat] = np.zeros((3,4))
            ########### SELECTION: CLUSTERS ############
                me1 = (T['cscRechitClusterNRechitChamberPlus11'].array()+ T['cscRechitClusterNRechitChamberPlus12'].array()+\
                T['cscRechitClusterNRechitChamberMinus11'].array()+ T['cscRechitClusterNRechitChamberMinus12'].array())
                # cluster = 
                #one cluster to pass trigger
                # cluster0 = np.logical_and(T['cscRechitClusterTimeWeighted'].array() < 12.5, T['cscRechitClusterTimeWeighted'].array() > -5)
                cluster = (T['cscRechitClusterTimeSpreadWeightedAll'].array()<25)
                cluster = cluster & (me1 ==0)
                
                if jetmetsys == 0: cluster = cluster & (T['cscRechitClusterJetVetoPt'].array()<JET_PT_CUT)
                elif jetmetsys == 1:cluster = cluster & (T['cscRechitClusterJetVetoPtJESUp'].array()<JET_PT_CUT)
                else: csccluster = cluster & (T['cscRechitClusterJetVetoPtJESDown'].array()<JET_PT_CUT)
                
                
                
                cluster = cluster & (np.abs(T['cscRechitClusterEta'].array()) < 1.9)
                cluster = cluster & np.logical_not((T['cscRechitClusterMuonVetoPt'].array() > 30) & T['cscRechitClusterMuonVetoGlobal'].array())    
            ########### SELECTION: EVENTS ############
                sel_ev[k] = T['HLT_CSCCSC'].array()
                # sel_ev  = np.logical_and(sel_ev,np.sum(cluster0, axis = 1) >= 1)
                sel_ev[k]  = np.logical_and(sel_ev[k],np.sum(cluster, axis = 1) == 2)

                cluster_pass_HLT = HLT_CSC(T['cscRechitClusterEta'].array(),T['cscRechitClusterNStation10'].array(),T['cscRechitClusterSize'].array())
                sel_ev[k]  = np.logical_and(sel_ev[k],np.sum(cluster_pass_HLT, axis = 1) >= 1) # at least one passes HLT
                sel_ev[k] = np.logical_and(sel_ev[k],T['nCscRechitClusters_nocut'].array()==2)
                sel_ev[k] = np.logical_and(sel_ev[k],T['nCscRings'].array()+T['nDtRings'].array()<10)
                sel_ev[k] = np.logical_and(sel_ev[k],T['jetVeto'].array())
                sel_ev[k] = np.logical_and(sel_ev[k],T['Flag_ecalBadCalibFilter'].array())
                sel_ev[k] = np.logical_and(sel_ev[k],T['Flag_all'].array())
                sel_ev[k] = sel_ev[k] & (np.sum(T['cscRechitClusterNStation10'].array(),axis=1) > 2)

                if cat == 'lowMET':
                    if jetmetsys == 0: sel_ev[k] = np.logical_and(sel_ev[k] ,T['Puppimet'].array() < 30)
                    elif jetmetsys == 1:sel_ev[k] = np.logical_and(sel_ev[k] ,T['PuppimetJESUp'].array() < 30)
                    else: sel_ev[k] = np.logical_and(sel_ev[k] ,T['PuppimetJESDown'].array() < 30)
                else:
                    if jetmetsys == 0: sel_ev[k] = np.logical_and(sel_ev[k] ,T['Puppimet'].array() >= 30)
                    elif jetmetsys == 1:sel_ev[k] = np.logical_and(sel_ev[k] ,T['PuppimetJESUp'].array() >= 30)
                    else: sel_ev[k] = np.logical_and(sel_ev[k] ,T['PuppimetJESDown'].array() >= 30)
                    
                    
                sel_ev[k] = np.logical_and(sel_ev[k] ,T['Puppimet'].array() < 200)

            ########### BRANCHES ############
               ##### event variables ##### 
                cscRechitClusterPhi0[k] = T['cscRechitClusterPhi'].array()[cluster][sel_ev[k]][:,0]
                cscRechitClusterPhi1[k] = T['cscRechitClusterPhi'].array()[cluster][sel_ev[k]][:,1]
                deltaPhi_cluster[k] = deltaPhi(np.array(cscRechitClusterPhi0[k]), np.array(cscRechitClusterPhi1[k]))
                cscRechitClusterDNN0[k] = T['cscRechitClusterDNN_bkgMC_plusBeamHalo'].array()[cluster][sel_ev[k]][:,0]
                cscRechitClusterDNN1[k] = T['cscRechitClusterDNN_bkgMC_plusBeamHalo'].array()[cluster][sel_ev[k]][:,1]
                cscRechitClusterTime0[k] = T['cscRechitClusterTimeWeighted'].array()[cluster][sel_ev[k]][:,0]
                cscRechitClusterTime1[k] = T['cscRechitClusterTimeWeighted'].array()[cluster][sel_ev[k]][:,1]
                cscRechitClusterSize0[k] = T['cscRechitClusterSize'].array()[cluster][sel_ev[k]][:,0]
                cscRechitClusterSize1[k] = T['cscRechitClusterSize'].array()[cluster][sel_ev[k]][:,1]
                cscRechitClusterEta[k] = T['cscRechitClusterEta'].array()[cluster][sel_ev[k]]
                cscRechitClusterNStation[k] = np.sum(T['cscRechitClusterNStation10'].array()[cluster][sel_ev[k]], axis=1)
                deltaR_cluster[k] = ((cscRechitClusterEta[k][:,0]-cscRechitClusterEta[k][:,1])**2 + deltaPhi_cluster[k]**2)**0.5

                if cat == 'lowMET': sel[k] = (np.abs(deltaPhi_cluster[k])>1.8) & (deltaR_cluster[k] < 3.5) 
                elif cat == 'highMET': sel[k] = (np.abs(deltaPhi_cluster[k])>0.4) & (deltaR_cluster[k] < 3.5)


                sel[k] = sel[k] & (cscRechitClusterDNN0[k] > 0.96) & (cscRechitClusterTime0[k]>-5) & (cscRechitClusterTime0[k]<12.5)
                if 'oot' in region:
                    sel[k] = sel[k] & (cscRechitClusterDNN1[k] > 0.96) & (cscRechitClusterTime1[k]<-12.5)
                elif 'invertDNN' in region:
                    sel[k] = sel[k] & (cscRechitClusterDNN1[k] < 0.96) & (cscRechitClusterDNN1[k] > 0.9) & (cscRechitClusterTime1[k]>-5) & (cscRechitClusterTime1[k]<12.5)
                elif 'signal' in region:
                    sel[k] = sel[k] & (cscRechitClusterDNN1[k] > 0.96) & (cscRechitClusterTime1[k]>-5) & (cscRechitClusterTime1[k]<12.5)
                else: assert(False)
                deltaPhi_cluster[k] = deltaPhi_cluster[k][sel[k]]
                cscRechitClusterSize1[k] = cscRechitClusterSize1[k][sel[k]]
                HMTEff[k] = T['cscRechitClusterHMTEfficiency'].array()[sel_ev[k]][sel[k]]

                cluster_pass_HLT = cluster_pass_HLT[sel_ev[k]][sel[k]]
                HMTEff[k] = 1-HMTEff[k][cluster_pass_HLT]
                HMTEff[k] = 1-np.prod(HMTEff[k],axis=1)



                if 'data' in k:
                    weight[k] = (T['weight'].array())[sel_ev[k]][sel[k]]*0.0 + 1                                                             
                    print(k, len(weight[k]))
                else: 
                    weight[k] = (T['pileupWeight'].array()*T['weight'].array())[sel_ev[k]][sel[k]]*HMTEff[k]*LUMI_SCALE
                    pileupWeightUp[k] = T['pileupWeightUp'].array()[sel_ev[k]][sel[k]]
                    pileupWeightDown[k] = T['pileupWeightDown'].array()[sel_ev[k]][sel[k]]
                    scaleWeights[k] = T['scaleWeights'].array()[sel_ev[k]][sel[k]]

                if cat == "lowMET": NHIT_CUT, DPHI_CUT = 180,2.8
                elif cat == "highMET": NHIT_CUT, DPHI_CUT = 160,2.3

                a = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)
                b = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
                c = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
                d = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)
                print(k, jetmetsys, np.sum(weight[k][a]),  np.sum(weight[k][b]),  np.sum(weight[k][c]),  np.sum(weight[k][d]))
                jetmet[cat][jetmetsys] = [np.sum(weight[k][a]),  np.sum(weight[k][b]),  np.sum(weight[k][c]),  np.sum(weight[k][d])]
                print(jetmet)
```

```python
def unc(up, nominal):
    ratio = np.sum(up)/np.sum(nominal)
    ratio_unc = (1/len(up) + 1/len(nominal))**0.5 * ratio
    return round(abs(1-ratio),4), round(ratio_unc,3)
```

## pileup uncertainty

```python
      
for cat in categories:
    
    if cat == "lowMET": NHIT_CUT, DPHI_CUT = 180,2.8
    elif cat == "highMET": NHIT_CUT, DPHI_CUT = 160,2.3
    for tree_k in tree.keys():
   
        k = f'signal_{tree_k}_{cat}'

        w = weight[k]
        wUp = weight[k]*pileupWeightUp[k]
        wDown = weight[k]*pileupWeightDown[k]

        
        a = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)
        b = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
        c = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
        d = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)

        if np.sum(w[a]) == 0:continue
        if len(w[a])<50:continue
        print(np.count_nonzero(a))
        print(k, 'a', (unc(wUp[a], w[a])[0] + unc(wDown[a], w[a])[0])/2)
        print(k, 'b', (unc(wUp[b], w[b])[0] + unc(wDown[b], w[b])[0])/2)
        print(k, 'c', (unc(wUp[c], w[c])[0] + unc(wDown[c], w[c])[0])/2)
        print(k, 'd', (unc(wUp[d], w[d])[0] + unc(wDown[d], w[d])[0])/2)
```

## scale weight

```python
sumofWeights = [1.1683354, 1.2007973,1.2275761,0.9691435,0.9999784,1.0245395,0.8174398,0.8460799,0.8683254]
for cat in categories:
    
    if cat == "lowMET": NHIT_CUT, DPHI_CUT = 180,2.8
    elif cat == "highMET": NHIT_CUT, DPHI_CUT = 160,2.3
    for tree_k in tree.keys():
   
        k = f'signal_{tree_k}_{cat}'

        w = weight[k]
        
        a = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)
        b = (cscRechitClusterSize1[k]>=NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
        c = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])<DPHI_CUT)
        d = (cscRechitClusterSize1[k]<NHIT_CUT) &  (np.abs(deltaPhi_cluster[k])>=DPHI_CUT)
        up = [0,0,0,0]
        down = [0,0,0,0]
            
        for i in [0,1,3,5,7,8]:
            
            w_temp = weight[k]*scaleWeights[k][:,i]/sumofWeights[i]
            # w_temp = weight[k]*scaleWeights[k][:,i]
            
            # down: 0,1,3
            # up: 5, 7, 8
            if i < 4:
                down[0] += (np.sum(w_temp[a])/np.sum(w[a])-1)**2
                down[1] += (np.sum(w_temp[b])/np.sum(w[b])-1)**2
                down[2] += (np.sum(w_temp[c])/np.sum(w[c])-1)**2
                down[3] += (np.sum(w_temp[d])/np.sum(w[d])-1)**2
            else: 
                up[0] += (np.sum(w_temp[a])/np.sum(w[a])-1)**2
                up[1] += (np.sum(w_temp[b])/np.sum(w[b])-1)**2
                up[2] += (np.sum(w_temp[c])/np.sum(w[c])-1)**2
                up[3] += (np.sum(w_temp[d])/np.sum(w[d])-1)**2
        up = np.array(up)**0.5
        down = np.array(down)**0.5
        print(tree_k,cat)
        print_str = ""
        for i in range(4):
            print_str += f"{up[i]*100:6.2f}\%(+) {100*down[i]:6.2f}\% (-) & "
        # print(k, np.round(100*np.array(up)**0.5,2), np.round(100*np.array(down)**0.5,2))
        print(print_str)
```
